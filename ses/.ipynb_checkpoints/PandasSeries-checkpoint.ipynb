```python
objSer = Series([4,2,-1,0])
print(objSer)
print(objSer.index)
```

```python
objSer.sum()
```

```python
objSerName = Series(['aryan', 'yash', 'meet', 'raj'])
print(objSerName)
print(objSerName.index)
```

```python
print(objSerName.sum())
print(objSerName.min())
print(objSerName.max())
```

```python
# label Series
labSer = Series([2,3,5,7,11,13],index = ["p1","p2","p3","p4","p5","p6"])
print(labSer)
print(labSer.index)
```

```python
print(objSer[0:2])
print(labSer.index[3:])
print(labSer["p1"])
```

```python
print(labSer[['p1','p4']])
print(objSer[[0,3,1]])
```

```python
print(f"Original Series: \n{objSer}")
print(f"Sum Series: {objSer.sum()}")
print(f"Mul Series: \n{objSer * 2}")
print(f"Sub Series: \n{objSer - 1}")
print(f"Min Series: {objSer.min()}")
print(f"Max Series: {objSer.max()}")
```

```python
mSub = {"Math": 70, "Eng": 75, "Guj": 79, "Sci": 69}
subSer = Series(mSub)
subSer
```

```python
subSer["Total"] = 0
subSer["Total"] = subSer.sum()
subSer
```

```python
subject = ["Math", "Eng", "Guj", "Grammer"]

subSer = Series(mSub, index=subject)
subSer
```

```python
subSer["Total"] = 0
subSer["Total"] = subSer.sum()
subSer
```

```python
subSer.notna()
```

```python
subSer.isna()
```

```python
subSer["Grammer"] = 67
subSer["Total"] = 0
subSer["Total"] = subSer.sum()
isna(subSer)
```

```python
notna(subSer)
```

```python
frdSub = Series({"Math":50, "Eng":55, "Computer":87})
frdSub
```

```python
merge = frdSub + subSer # NaN + any value = NaN
merge
```
