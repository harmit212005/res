```python
import matplotlib.pyplot as plt
from pandas import Series,DataFrame,isna,notna
```

```python
students = DataFrame({"Name":["Jay","Karan","Jinal","Dipa","Mona","Krupesh"],
            "Id":[1,2,3,4,5,6],
            "Gender":("Male","Male","Female","Female","Female","Male"),
            "DA":(76,89,76,54,38,89),
            "Project":(56,75,93,78,76,90),
            "ADBMS":(78,87,56,88,64,89),
            "Course":("MCA","Msc.IT","MCA","Msc.IT","Msc.IT","Msc.IT")
        })
students
```

```python
students["Total"] = students.loc[:,["DA", "Project", "ADBMS"]].sum(axis=1)
students
```

```python
x = students["Name"]
y = students["Total"]

plt.bar(x,y,label="Student Data",color='red',width=0.5)
plt.ylabel("Total Marks")
plt.xlabel("Student Name")
plt.title("Student Marks")
plt.legend(loc='upper center')
```

```python
students["Per"] = students.apply(lambda row: row["Total"] / 3 if row["DA"] > 40 and 
                                      row["Project"] > 40 and row["ADBMS"] > 40 else 0, axis=1)
students
```

```python
x = students["Name"]
y = students["Per"]

plt.bar(x,y,label="Student Percentage",color='red',width=0.5)
plt.ylabel("Total Percentage")
plt.xlabel("Student Name")
plt.title("Student Percentage")
plt.legend(loc='upper center')
```

```python
maleStudent = students.loc[students["Gender"] == "Male"]
femaleStudent = students.loc[students["Gender"] == "Female"]

x = maleStudent["Name"]
y = maleStudent["Total"]

x1 = femaleStudent["Name"]
y1 = femaleStudent["Total"]

plt.bar(x,y,label="Male Students", color='green',width=0.5)
plt.bar(x1,y1,label="Female Students", color='blue',width=0.5)
plt.ylabel("Total Marks")
plt.xlabel("Student Name")
plt.title("Student Marks")
plt.legend()
```

```python
x = students["Name"]
y = students["DA"]
y1 = students["Project"]

plt.plot(x,y,'green')
plt.plot(x,y1,'blue')
plt.ylabel("Total Marks")
plt.xlabel("Student Name")
plt.title("Student Marks")
plt.show()
```

```python
import numpy as np

x = students["Name"]
y = students["DA"]
y1 = students["Project"]

br1 = np.arange(len(y))
br2 = [r + 0.25 for r in br1]
 
plt.bar(br1,y,width=0.25,label="DA",color="c")
plt.bar(br2,y1,width=0.25,label="Project",color="b")

plt.xticks([r + 0.25 / 2 for r in br1], x)
plt.ylabel("Marks")
plt.xlabel("Student Name")
plt.title("Student Marks")
plt.legend()
plt.show()
```

```python
students['Total'].plot(kind='pie', autopct='%1.1f%%', startangle=180)

plt.title('Total Score Contribution by Student')
plt.ylabel('Student marks')
plt.legend()
```

```python
import numpy as np
x = np.arange(len(students["Name"]));
width = 0.25

daBar = plt.bar(x-width-0.02, students["DA"], width, label="DA",color="c")
projectBar = plt.bar(x, students["Project"], width, label="Project",color="orange")
adbmsBar = plt.bar(x+width+0.02, students["ADBMS"], width, label="ADBMS",color="green")
plt.bar_label(daBar, padding=3)
plt.bar_label(projectBar, padding=3)
plt.bar_label(adbmsBar, padding=3)
plt.legend(
    title='Subjects',
    loc='upper center',
    # fontsize=10,
    # frameon=True,
    # facecolor='#f8f9fa',
    # edgecolor='gray',
    ncol=3
)

plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('DA vs Project vs ADBMS marks')
plt.xticks(x,students["Name"])
plt.ylim(0,120)
```

```python
months = ["January", "February", "March", "April", "May", "June", "July"]
sales = [65, 72, 68, 85, 78, 92, 88]

maximum = max(sales)
minimum = min(sales)

maxIndex = sales.index(maximum)
minIndex = sales.index(minimum)

max_month = months[maxIndex]
min_month = months[minIndex]

plt.plot(months, sales, marker='o', linestyle='-.', markersize=5, label="Sales", color='b',
         markerfacecolor='k', markeredgecolor='k')
plt.scatter(max_month, maximum, s=100,label='Maximum', color='g', marker='o', alpha=0.75, edgecolors='g')
plt.scatter(min_month, minimum, s=100,label='Minimum', color='r', marker='o', alpha=0.75, edgecolors='r')

plt.title("Employee monthly sales performance")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)
plt.legend()

print(f"Maximum value is {maximum} in the month of {max_month}")
print(f"Maximum value is {minimum} in the month of {min_month}")
```
