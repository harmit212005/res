```python
# 27. Take a matrix input from the user and count how many elements are even and how many are odd using NumPy.

row = int(input("Enter no of rows"))
col = int(input("Enter no of coluns"))
arr = np.array([int(input()) for _ in range(row*col)]).reshape(row,col)

print("Total even no:",np.sum(arr%2==0))
print("Total odd no:",np.sum(arr%2!=0))
```

```python
# 31. Write a program to accept a 2D array from the user and find the sum of each row and each column separately.

row = int(input("Enter rows"))
col = int(input("Enter column"))
arr = np.array([int(input("Enter elements")) for _ in range(row*col)]).reshape(row,col)

print("Original array:\n",arr)
print("Column wise sum:",arr.sum(axis=1))
print("Row wise sum:",arr.sum(axis=0))
```

```python
# 32. Accept a list of numbers from the user and find the mean, median, and mode.

from statistics import mode

arr = np.array(list(map(int,input("Enter numbers with space").split())))
print("Mean:",np.mean(arr))
print("Median:",np.median(arr))
print("Mode:",mode(arr))
```

```python
# 33. Write a program to generate 50 random numbers between 1 and 100 and calculate:
# o Mean
# o Median
# o Standard deviation
# o Variance
# o Weighted average

arr = np.random.randint(1,101,50)

print("Mean:",np.mean(arr))
print("Median:",np.median(arr))
print("Devication:",np.std(arr))
print("Variance:",np.var(arr))
print("Average:",np.average(arr))
```

```python
# 34. Create a 2D array using user input and find the mean of each row and each column separately.

row = int(input("Enter range for row"))
col = int(input("Enter range for col"))
arr = np.array([int(input("Enter element")) for _ in range(row*col)]).reshape(row,col)

print("Original:\n",arr)
print("Column wise Mean:",np.mean(arr,axis=1))
print("Row wise Mean:",np.mean(arr,axis=0))
```

```python
# 35. Accept temperature data for a week and calculate:
# o Average temperature
# o Highest and lowest temperatures
# o Days with temperature above the weekly average

temp = np.array([float(input("Enter Temperature")) for _ in range(7)])
a = temp.mean()

print("Average Temp:",a)
print("Maximum Temp:",temp.max())
print("Minimum Temp:",temp.min())
print("Days with temperature above the weekly average:",temp[temp>a])
```

```python
# 36. Generate marks of 200 students randomly.
# o Perform all of the following:
# o Sort marks
# o Remove duplicates
# o Find topper
# o Find average marks
# o Find pass percentage
# o Find distinction percentage
# o Find students scoring above average
# o Create grade distribution:
#  A (≥90)
#  B (75–89)
#  C (50–74)
#  D (40–49)
#  F (<40)
# o Display count of students in each grade.
# o Plot a histogram of marks using Matplotlib.

import matplotlib.pyplot as plt

marks = np.random.randint(0, 101, 200)

print("Sorted Marks:\n",np.sort(marks))

print("\nUnique Marks:\n",np.unique(marks))

print("\nTopper Marks: ", np.max(marks))

average = np.mean(marks)
print("Average Marks:", average)

passper = np.sum(marks >= 40) / len(marks) * 100
print("Pass Percentage:", passper)

distinctionper = np.sum(marks >= 75) / len(marks) * 100
print("Distinction Percentage:", distinctionper)

print("\nStudents Above Average:\n",marks[marks > average])

print("\nGrade Distribution")
print("Total A grade studens:", np.sum(marks >= 90))
print("Total B grade studens:", np.sum((marks >= 75) & (marks <= 89)))
print("Total C grade studens:", np.sum((marks >= 50) & (marks <= 74)))
print("Total D grade studens:", np.sum((marks >= 40) & (marks <= 49)))
print("Total F grade studens:", np.sum(marks >= 40))
```
