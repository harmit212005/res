```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (mean_absolute_error, mean_squared_error, r2_score)
```

```python
housing = fetch_california_housing(as_frame = True)
df = housing.frame
print("Dataset shape:",df.shape)
df.head()
```

```python
# Know a dataset
print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDublicate rows:")
print(df.duplicated().sum())

print("\nStatistical summary:")
df.describe()
```

```python
df_dirty = df.copy()

df_dirty.loc[10, "MedInc"] = np.nan

df_dirty.loc[20, "HouseAge"] = np.nan

df_dirty.loc[30, "AveRooms"] = np.nan

df_dirty.loc[40, "Population"] = np.nan

print(df_dirty.isnull().sum())
```

```python
X = df_dirty[
    [
        "MedInc",
        "HouseAge",
        "AveRooms",
        "AveBedrms",
        "Population",
        "AveOccup",
        "Latitude",
        "Longitude"
    ]
]

y = df_dirty["MedHouseVal"]

# X = df_dirty.iloc[:, :-1] 
# y = df_dirty.iloc[:, -1]

print("\nInput Fature (X):")
print(X.columns)

print("\nTarget (y)")
print(y.name)

print("\nShape of X:")
print(X.shape)

print("\nShape of y:")
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("\nTraining data shape:")
print(X_train.shape)
print("\nTesting data shape:")
print(X_test.shape)
```

```python
imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

```python
# Fourmula: z = (X - mean) / standard_deviation

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)
```

```python
model = LinearRegression()

# The model learns the relationship between:
# X_train_scaled -> y_train

# In mathematical form:
# y = b0 + b1X1 + b2X2 + ... + b8X8
model.fit(X_train_scaled, y_train)
```

```python
# intercept_ represents b0 in the equation
print("\nIntercept (b0):")
print(model.intercept_)
```
