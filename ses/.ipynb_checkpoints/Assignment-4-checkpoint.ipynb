```python
# Classify iris flowers (Decision Tree)
# Objective: Predict a flower species using sepal and petal measurements.
# Dataset: load_iris()
# Algorithm: Decision Tree Classifier
# The Iris dataset contains 150 flower samples, four numerical input features, and three species classes.
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier, plot_tree

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
iris = load_iris(as_frame=True)
df = iris.frame

print("Dataset Shape:", df.shape)
df.head()
```

```python
print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
df.describe()
```

```python
print("Species Class Count:")
print(df["target"].value_counts().sort_index())

print("\nSpecies Names:")
print(iris.target_names)
```

```python
X = df.iloc[:,:-1]
y = df["target"]

print("X Shape:", X.shape)
print("\ny Shape:", y.shape)

print("\nInput Features:")
print(X.columns.tolist())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

```python
imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

```python
model = DecisionTreeClassifier(random_state=42)

model.fit(X_train_imputed, y_train)
```

```python
y_pred = model.predict(X_test_imputed)

print("Actual Values:")
print(y_test.values[:10])

print("\nPredicted Values:")
print(y_pred[:10])
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
print("Classification Report:")
print(classification_report(
    y_test, y_pred, target_names=iris.target_names
))
```

```python
cm = confusion_matrix(y_test, y_pred)

sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.xlabel("Predicted Species")
plt.ylabel("Actual Species")
plt.title("Decision Tree - Confusion Matrix")
plt.show()
```

```python
plt.figure(figsize=(14, 8))
plot_tree(
    model,
    feature_names=iris.feature_names,
    class_names=iris.target_names,
    filled=True,
    rounded=True
)
plt.title("Decision Tree Classifier")
plt.show()
```

```python
# K-Nearest Neighbors (KNN) Using the Iris Dataset
# Aim: To classify an iris flower into one of three species using KNN.
# Dataset: Scikit-learn built-in Iris dataset
# Input features: Sepal length, sepal width, petal length, petal width
# Target: Setosa, Versicolor, or Virginica
# Algorithm: KNN Classifier
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
iris = load_iris(as_frame=True)
df = iris.frame

print("Dataset Shape:", df.shape)
df.head()
```

```python
print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
df.describe()
```

```python
X = df.iloc[:,:-1]
y = df["target"]

print("X Shape:", X.shape)
print("\ny Shape:", y.shape)

print("\nSpecies Names:")
print(iris.target_names)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

```python
imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train_scaled, y_train)
```

```python
y_pred = knn_model.predict(X_test_scaled)

print("Actual Values:")
print(y_test.values[:10])

print("\nPredicted Values:")
print(y_pred[:10])
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
print("Classification Report:")
print(classification_report(
    y_test, y_pred, target_names=iris.target_names
))
```

```python
new_flower = pd.DataFrame(
    [[5.1, 3.5, 1.4, 0.2]],
    columns=iris.feature_names
)

new_flower_imputed = imputer.transform(new_flower)
new_flower_scaled = scaler.transform(new_flower_imputed)

prediction = knn_model.predict(new_flower_scaled)[0]

print("Predicted Species:", iris.target_names[prediction])
```

```python
cm = confusion_matrix(y_test, y_pred)

sns.heatmap(
    cm, annot=True, fmt="d", cmap="Blues",
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.xlabel("Predicted Species")
plt.ylabel("Actual Species")
plt.title("KNN - Confusion Matrix")
plt.show()
```
