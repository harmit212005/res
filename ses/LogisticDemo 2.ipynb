```python
features = [
    "pclass",
    "sex",
    "age",
    "sibsp",
    "parch",
    "fare",
    "embarked"
]

X = df[features] 
y = df["survived"]
```

```python
print("X Shape:", X.shape)
print("\ny Shape:", y.shape)
```

```python
numeric_features = [
    "pclass",
    "age",
    "sibsp",
    "parch",
    "fare"
]

categorical_features = [
    "sex",
    "embarked"
]
```

```python
df[(df["survived"] == 1) & (df["alone"] == True)]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.20, random_state = 42, stratify = y
)
```

```python
print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
print("\nTraining Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)
```

```python
numeric_imputer = SimpleImputer(strategy="median")
numeric_scaler = StandardScaler()

numeric_pipeline = Pipeline(
    steps=[
        ("imputer", numeric_imputer),
        ("scaler", numeric_scaler)
    ]
)
```

```python
categorical_imputer = SimpleImputer(strategy="most_frequent")
categorical_encoder = OneHotEncoder(handle_unknown="ignore")

categorical_pipeline = Pipeline(
    steps=[
        ("imputer", categorical_imputer),
        ("encoder", categorical_encoder)
    ]
)
```

```python
# Combine Numerical and Categorical Preprocessing

preprocessor = ColumnTransformer(
    transformers = [
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features)
    ]
)
```

```python
logistic_model = LogisticRegression(max_iter=1000)
```

```python
model = Pipeline(
    steps = [
        ("preprocesser", preprocessor),
        ("classifier", logistic_model)
    ]
)
```

```python
# Train the logistic regression model

model.fit(X_train, y_train)
```

```python
# Make Class Prediction

y_pred = model.predict(X_test)
```

```python
# Grt predict probability

# Column 0 = probablity of class 0.
# Column 1 = probablity of class 1.

y_probability = model.predict_proba(X_test)

# Extract the probability of survival
# [:,1] means we select the probability of class 1.
y_probability_positive = y_probability[:,1]
```

```python
# Display first 10 prediction

# Display the first 10 actual values
print("Actual Values:")
print(y_test.iloc[:10].values)

# Display the first 10 predicted values
print("\nPredicted Values:")
print(y_pred[:10])

# Display the first 10 predicted probabilities.
print("\nProbability of Survival:")
print(y_probability_positive[:10])
```

```python
# Create Result Dataframe

results = pd.DataFrame({
    "Actual" : y_test.values,
    "Predicted" : y_pred,
    "Survival_Probability" : y_probability_positive
})

print("Prediction Results:")
results.head(10)
```

```python
# Calculate Accuracy

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```
