```python
# question
# Multiple Line Charts 
# The following are the dataset 
    # months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"] 
    # product_A = [200, 250, 300, 280, 350, 400] 
    # product_B = [180, 220, 270, 320, 370, 390] 

# Tasks 
    # Plot both products on the same graph. 
    # Use different colors and markers. 
    # Add a legend. 
    # Add grid lines. 
    # Identify which product performed better overall. 

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_A = [200, 250, 300, 280, 350, 400]
product_B = [180, 220, 270, 320, 370, 390]

plt.plot(months, product_A, color="b", marker="o", label="Product A")
plt.plot(months, product_B, color="r", marker="x", label="Product B")

plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Product A vs Product B Performance")

plt.legend()
plt.grid(True)

total_A = sum(product_A)
total_B = sum(product_B)

print(f"Total sales of Product A: {total_A}")
print(f"Total sales of Product B: {total_B}")

if total_A > total_B:
    print("Product A performed better overall.")
elif total_B > total_A:
    print("Product B performed better overall.")
else:
    print("Both products performed equally overall.")
```

```python
# Question
# Horizontal Bar Chart 
#     The following are the dataset. 
#     employees = ["John", "Emma", "David", "Sophia", "Michael"] 
#     tasks_completed = [45, 60, 38, 72, 55] 
# Task 
#     Create a horizontal bar chart. 
#     Sort employees according to completed tasks. 
#     Highlight the most productive employee. 
#     Add chart title and labels. 

import numpy as np

employees = ["John", "Emma", "David", "Sophia", "Michael"] 
tasks_completed = [45, 60, 38, 72, 55]

sorted_index = np.argsort(tasks_completed)

sorted_emp = np.array(employees)[sorted_index]
sorted_tasks = np.array(tasks_completed)[sorted_index]

plt.barh(sorted_emp, sorted_tasks, color='c')

plt.title("Tasks Completed by Employees")
plt.xlabel("Tasks Completed")
plt.ylabel("Employees")
```

```python
# Question
# Quarterly Revenue Analysis 

#     The following are the dataset. 
#         quarters = ["Q1", "Q2", "Q3", "Q4"] 
#         Take the value of quarter revenue from the user. 
    
#     Task 
#         Plot a bar chart for quarterly revenue. 
#         Add revenue values on top of bars. 
#         Format the Y-axis with currency labels. 
#         Determine the quarter with maximum growth. 

quarters = ["Q1", "Q2", "Q3", "Q4"]
revenue = []

for q in quarters:
    value = int(input("Enter revenue for " + q + ": "))
    revenue.append(value)

x = np.arange(len(quarters))
width = 0.5

revenueBar = plt.bar(x, revenue, width, label="Revenue", color="skyblue")

plt.bar_label(revenueBar, padding=3)
plt.xlabel("Quarter")
plt.ylabel("Revenue")
plt.title("Quarterly Revenue Analysis")
plt.xticks(x, quarters)

max_revenue = np.max(revenue)
y_values = np.linspace(0, max_revenue, 6)

plt.yticks(y_values, ["$" + str(int(value)) for value in y_values])
plt.grid(axis="y")


growth = []

for i in range(1, len(revenue)):
    growth_value = revenue[i] - revenue[i - 1]
    growth.append(growth_value)

max_growth = np.max(growth)
position = growth.index(max_growth)
max_growth_quarter = quarters[position + 1]

print("Quarter with maximum growth:", max_growth_quarter)
print("Maximum growth:", int(max_growth))
```

```python
# question
# Monthly Expense Tracker 
#     The following are the dataset. 
#         categories = ["Rent", "Food", "Transport", "Utilities", "Entertainment"] 
#         expenses = [15000, 7000, 3000, 2500, 2000]  
#     Task 
#         Create a horizontal bar chart.         
#         Display expense amounts on bars.         
#         Identify the highest expense category.         
#         Calculate total monthly expenses. 

categories = ["Rent", "Food", "Transport", "Utilities", "Entertainment"]
expenses = [15000, 7000, 3000, 2500, 2000]

y = np.arange(len(categories))
height = 0.5

expenseBar = plt.barh(y, expenses, height, label="Expenses", color="skyblue")

plt.bar_label(expenseBar, padding=3)
plt.ylabel("Category")
plt.xlabel("Expense")
plt.title("Monthly Expense Tracker")
plt.yticks(y, categories)

max_expense = np.max(expenses)

x_values = np.linspace(0, max_expense, 6)

plt.xticks(x_values, ["₹" + str(int(value)) for value in x_values])
plt.grid(axis="x")

max_expense_value = np.max(expenses)
position = expenses.index(max_expense_value)

highest_category = categories[position]

total_expenses = np.sum(expenses)

print("Highest expense category:", highest_category)
print("Highest expense:", int(max_expense_value))
print("Total monthly expenses:", int(total_expenses))
```
