```python
if df.duplicated().sum() != 0:
    df = df.drop_duplicates()
else:
    print("The dataset does not contains dublicate value.")
```

```python
# consistent data
print(f"unique season: {df['season'].unique()}")
print(f"unique month: {df['mnth'].unique()}")
print(f"unique year: {df['yr'].unique()}")
print(f"unique weekday: {df['weekday'].unique()}")
```

```python
print(df["dteday"].dtype)
df["dteday"] = pd.to_datetime(df['dteday'])
print(df["dteday"].dtype)
```

```python
selected_columns = ["dteday","season","yr","mnth","weekday","workingday","weathersit","temp","hum","windspeed","cnt"]
analysis_df = df[selected_columns]
print(analysis_df.columns)
```

```python
def month_map(value):
    dict = {1:"JAN", 2:"FEB", 3:"MAR", 4:"APR", 5:"MAY", 6:"JUN", 7:"JUL", 8:"AUG", 9:"SEP", 10:"OCT", 11:"NOV", 12:"DEC"}
    return dict[value]
```

```python
analysis_df["year"] = analysis_df['yr'].apply(lambda year: 2011 if year == 0 else 2012)

analysis_df["month"] = analysis_df['mnth'].apply(month_map)

print(analysis_df["month"].unique())
print(analysis_df["year"].unique())
```

```python
# Aggregation

total_demand = analysis_df["cnt"].sum()
print(f"Total demand : {total_demand}")

year_sum_demand = analysis_df.groupby("year")["cnt"].sum()
print(f"Year wise demand :\n {year_sum_demand}")

yr_demand = analysis_df.groupby("year")["cnt"].mean()
print(f"Year wise average demand :\n {yr_demand}")

yr_mnth_demand = analysis_df.groupby(["year","month"])["cnt"].mean()
print(f"Monthly average bike demand by year and month wise :\n {yr_mnth_demand}")

mnth_demand = analysis_df.groupby("month")["cnt"].mean()
print(f"Monthly average bike demand by month wise :\n {mnth_demand}")
```

```python
print(f'Total demand of year 2011:\n {analysis_df.loc[analysis_df["year"] == 2011,["cnt"]].sum()}')
print(f'Total active days of year 2011:\n {analysis_df.loc[analysis_df["year"] == 2011,["cnt"]].count()}')

avg_2011 = analysis_df.loc[analysis_df["year"] == 2011,["cnt"]].sum() / analysis_df.loc[analysis_df["year"] == 2011,["cnt"]].count()
print(f"Average demand of year 2011:\n {avg_2011} ")
print(f'Total demand of April month\n {analysis_df.loc[analysis_df["month"] == "APR",["cnt"]].sum()}')
print(f'Total active days of April month\n {analysis_df.loc[analysis_df["month"] == "APR",["cnt"]].count()}')

april_avg=analysis_df.loc[analysis_df["month"] == "APR",["cnt"]].sum() / analysis_df.loc[analysis_df["month"] == "APR",["cnt"]].count()
print(f"Average demand of April Month\n:{april_avg}")
```

```python
import matplotlib.pyplot as plt
 
data_2011 = analysis_df[analysis_df["year"] == 2011]
 
plt.plot(data_2011.index, data_2011["cnt"], color="k", label="Demand")
 
plt.title("Demand in 2011")
plt.xlabel("Days")
plt.ylabel("Demand")
plt.grid(True)
plt.legend()
```
